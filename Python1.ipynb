## What are data structures?
A data structure is a way of organising and storing data in a computer so that we can access, manage and work with it efficiently.

The four Python data structures we will learn
1. List
2. Tuple
3. Dictionary
4. Set

## Understanding the Four Data Structures

A. **Lists**— Storing multiple items in one variable
A list is a data structure that allows us to store multiple items in a single variable.

### Characteristics of List
1. Lists use square brackets []
2. Items are stored in an order.
3. Items can be changed.
4. Duplicate items are allowed.
5. Items can be accessed using their positions, called indexes


B. **Tuples** — Storing information that should not change
A tuple is similar to a list because it stores multiple items in an ordered collection.
The major difference is that a tuple cannot be changed after it has been created.

### Characteristics of Tuples
1. Tuples use parentheses ().
2. Items are ordered.
3. Duplicates are allowed.
4. Items can be accessed using indexes.
5. Items cannot be changed after creation.

C. **Dictionaries** — Storing information as key-value pairs
This is particularly important for data analysis.
A dictionary stores information using keys and values.

### Characteristics of Dictionaries
1. Dictionaries use curly brackets {}.
2. Information is stored as key-value pairs.
3. Keys must be unique.
4. Values can be changed.
5. You access information using its key.


D. **Sets** — Storing unique items
A set is a collection of items that does not allow duplicates.

### Characteristics of Sets
1. Sets use curly brackets {} when created directly.
2. They do not allow duplicate values.
3. They do not guarantee a particular order.
4. They can be updated.
5. They do not support indexing.

## Formating Outputs using F-String

 F-strings is used to insert variables directly into a text string. The f is places before the opening quotation mark and variables are placed inside a {}

```python
# example
name = 'Adamu'
age = 30
print(name)
print(age)

#printing using f-strings
print(f'My name is {name} and I am {age} years old. Nice to meet you!')
```

```python
# f-string and no f-string

#without f-string; is prone to error
food = 'Yam'
price = 7000
print('I was at the market the other day, and the price of', food, 'is', price,'.')

#with f-string; is not prone to error
print(f'I was at the market the other day, and the price of {food} is {price}.')
```

```python
# Calculation inside f-strings
quantity = 5
price = 6500

print(f'The total cost of the item is \u20A6{quantity*price}')
```

```python
# Formatting numbers properly using f-string
print(f'The total cost of the item is \u20A6{quantity*price:,}')

sales = 7000000
print(f'Sales: \u20A6{sales:,.2f}')
```

## What are conditional statements?
Conditional statements allow Python to make decisions based on whether a particular condition is true or false.

The three main keywords are:
- if
- elif
- else

B. Understanding the if statement
An if statement tells Python to execute a particular instruction when a condition is true.

C. Understanding else
What if we want Python to display another message when the condition is false?
We use else.

D. Understanding elif
Sometimes we have more than two possible outcomes.
We can use elif to check additional conditions.



```python
# if statement
age = 25
if age >= 18:
    print('You are eligible to vote')
```

## Logical operators: and, or, not
Logical operators allow us to combine conditions.

1. and — Both conditions must be true.

Example: A customer qualifies for a special offer if they spend at least ₦100,000 and are registered members.

2. or — At least one condition must be true.

Example: A customer receives free delivery if they spend at least ₦100,000 or live in Lagos.

3. not — Reverses a Boolean condition.

### Introducing Pandas
Imagine that a supermarket has 10,000 sales records. Each record contains a product name, quantity sold, price, date and customer information.

We could store some of this information using lists and dictionaries, as we learnt in our previous class. But managing thousands of records manually would become difficult.

This is where Pandas comes in.

**Pandas** is a **Python library** designed to help us organise, manipulate, analyse and work with structured data."

What is a library?

A library is a collection of pre-written code that provides functions and tools for performing specific tasks.

Instead of writing everything from scratch, we can import a library and use the tools it provides.

Libraries and their	common use;

**Pandas**- Data manipulation and analysis

**NumPy**- Numerical calculations

**Matplotlib**- Data visualisation

**Seaborn**- 	Statistical data visualisation

### What is a DataFrame?
A DataFrame is a two-dimensional data structure in Pandas that organises information into rows and columns.

Note;
- A row represents one record or observation.
- A column represents a particular variable or attribute.
- A cell contains an individual value.

For example:
- Each row represents a product's sales record.
- The Product column contains product names.
- The Quantity column contains quantities sold.
- The Price column contains product prices.
